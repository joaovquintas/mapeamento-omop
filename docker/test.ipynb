```python
from pymilvus import connections, utility, Collection
from sentence_transformers import SentenceTransformer

MILVUS_URI = "http://localhost:19530"
COLLECTION = "devices_semantic"
EMBEDDING_MODEL = "MohammadKhodadad/MedTE-cl15-step-8000"

connections.connect("default", uri=MILVUS_URI)
print("existe?", utility.has_collection(COLLECTION))

col = Collection(COLLECTION)
print("entities:", col.num_entities)

col.load()

model = SentenceTransformer(EMBEDDING_MODEL, device="cuda")
q = "passage: arteriovenous fistula"
vec = model.encode([q], convert_to_numpy=True, normalize_embeddings=True)[0].astype("float32")

hits = col.search(
    data=[vec.tolist()],
    anns_field="name_embedding",
    param={"metric_type": "IP", "params": {"ef": 64}},
    limit=10,
    output_fields=["concept_id", "concept_name"],
)

for hit in hits[0]:
    print(hit.distance, hit.entity.get("concept_id"), hit.entity.get("concept_name"))

col.release()
```
